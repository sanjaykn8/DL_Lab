```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import dnnlib
import legacy

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

network_pkl = "https://nvlabs-fi-cdn.nvidia.com/stylegan2-ada-pytorch/pretrained/ffhq.pkl"

with dnnlib.util.open_url(network_pkl) as f:
    G = legacy.load_network_pkl(f)['G_ema'].to(device)

print("Model loaded")
```

```python
def generate_image(ws):
    img = G.synthesis(ws, noise_mode='const')
    img = (img.clamp(-1, 1) + 1) * 127.5
    img = img.permute(0, 2, 3, 1).to(torch.uint8).cpu().numpy()
    return img[0]

def show_images(images, cols=7):
    plt.figure(figsize=(15,5))
    for i, img in enumerate(images):
        plt.subplot(1, cols, i+1)
        plt.imshow(img)
        plt.axis('off')
    plt.show()

def lerp(a, b, t):
    return a + (b - a) * t

# Ensure W+ shape
def to_wplus(w):
    if w.dim() == 2:
        w = w.unsqueeze(1).repeat(1, G.num_ws, 1)
    assert w.shape[1] == G.num_ws
    return w
```

```python
z1 = torch.randn(1, G.z_dim).to(device)
z2 = torch.randn(1, G.z_dim).to(device)

# --- Z space interpolation ---
z_images = []
for t in np.linspace(0, 1, 7):
    z = lerp(z1, z2, t)
    w = G.mapping(z, None)
    ws = to_wplus(w)
    z_images.append(generate_image(ws))

show_images(z_images)
```

```python
w1 = G.mapping(z1, None)
w2 = G.mapping(z2, None)

ws1 = to_wplus(w1)
ws2 = to_wplus(w2)

w_images = []
for t in np.linspace(0, 1, 7):
    ws = lerp(ws1, ws2, t)
    w_images.append(generate_image(ws))

show_images(w_images)
```

```python
z3 = torch.randn(1, G.z_dim).to(device)

w3 = G.mapping(z3, None)

ws3 = to_wplus(w3)

# mix ws1 and ws3
mixed_ws = ws1.clone()
mixed_ws[:, 8:, :] = ws3[:, 8:, :]

img1 = generate_image(ws1)
img2 = generate_image(ws3)
mixed_img = generate_image(mixed_ws)

show_images([img1, img2, mixed_img], cols=3)
```

```python
z_samples = torch.randn(10000, G.z_dim).to(device)
w_samples = G.mapping(z_samples, None)
w_avg = w_samples.mean(dim=0, keepdim=True)

psi_values = [1.0, 0.7, 0.5, 0.3]
trunc_images = []

for psi in psi_values:
    w_trunc = w_avg + psi * (w1 - w_avg)
    ws_trunc = to_wplus(w_trunc)
    trunc_images.append(generate_image(ws_trunc))

show_images(trunc_images, cols=len(psi_values))
```
