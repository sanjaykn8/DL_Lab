```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
    return 1.0 / (1+np.exp(-x))

def dsigmoid(x):
    return sigmoid(x) * (1 - sigmoid(x))
```

```python
def tanh(x):
    return np.tanh(x)

def dtanh(x):
    return 1 - np.tanh(x)**2
```

```python
def relu(x):
    return np.maximum(0, x)

def drelu(x):
    return (x > 0).astype(float)
```

```python
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def dleaky_relu(x, alpha=0.01):
    dx = np.ones_like(x)
    dx[x < 0] = alpha
    return dx
```

```python
def softmax(z):
    z = np.asarray(z, dtype=np.float64)
    z = z - np.max(z) 
    expz = np.exp(z)
    return expz / np.sum(expz)
```

```python
def plots():
    x = np.linspace(-10, 10, 1000)

    sig = sigmoid(x); d_sig = dsigmoid(x)
    th = tanh(x); d_th = dtanh(x)
    r = relu(x); d_r = drelu(x)
    lr = leaky_relu(x, alpha=0.01); d_lr = dleaky_relu(x, alpha=0.01)

    fig = plt.figure(constrained_layout=True, figsize=(12, 10))
    gs = fig.add_gridspec(3, 2)

    # Sigmoid
    ax1 = fig.add_subplot(gs[0, 0])
    ax1.plot(x, sig, label='sigmoid', color='C0')
    ax1.plot(x, d_sig, '--', label="sigmoid'", color='C0', alpha=0.7)
    ax1.set_title("Sigmoid and derivative")
    ax1.legend()

    # Tanh
    ax2 = fig.add_subplot(gs[0, 1])
    ax2.plot(x, th, label='tanh', color='C1')
    ax2.plot(x, d_th, '--', label="tanh'", color='C1', alpha=0.7)
    ax2.set_title("Tanh and derivative")
    ax2.legend()

    # ReLU
    ax3 = fig.add_subplot(gs[1, 0])
    ax3.plot(x, r, label='ReLU', color='C2')
    ax3.plot(x, d_r, '--', label="ReLU'", color='C2', alpha=0.7)
    ax3.set_title("ReLU and derivative")
    ax3.set_ylim(-0.5, 10) 
    ax3.legend()

    # Leaky ReLU
    ax4 = fig.add_subplot(gs[1, 1])
    ax4.plot(x, lr, label='Leaky ReLU', color='C3')
    ax4.plot(x, d_lr, '--', label="Leaky ReLU'", color='C3', alpha=0.7)
    ax4.set_title("Leaky ReLU and derivative")
    ax4.legend()
```

```python
plots()
```

```python
logits = np.array([1.0, 0.0, -1.0])
print("softmax =", softmax(logits))
```

```python
plt.plot(logits, softmax(logits), label = "softmax")
plt.legend()
```
