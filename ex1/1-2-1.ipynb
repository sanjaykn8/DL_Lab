```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def func_1(x):
    return 3 * x**2 + 12 * x + 7

def func_2(x):
    return 2 * x**4 - 8 * x**2 + 5
```

```python
def check_curvature(f, points, h=1e-4):
    vals = np.array([
        (f(x + h) - 2 * f(x) + f(x - h)) / (h * h)
        for x in points
    ])

    if np.all(vals > 0):
        return "Strictly Convex"
    if np.all(vals < 0):
        return "Strictly Concave"
    if np.any(vals > 0) and np.any(vals < 0):
        return "Non-Convex"
    
    return "Flat"
```

```python
points = np.linspace(-5, 5, 201)
print("Curvature f1:", check_curvature(func_1, points))
print("Curvature f2:", check_curvature(func_2, points))
```

```python
def f1(x): return 3*x**2 + 12*x + 7
def f2(x): return x**2 - 4*x + 1
def f3(x): return 0.5*x**2 + 2*x + 3

def g1(x): return 2*x**4 - 8*x**2 + 5
def g2(x): return x**4 - x**2
def g3(x): return -x**4 + 4*x**2
```

```python
points = np.linspace(-5, 5, 400)
```

```python
plt.plot(points, f1(points), label="Convex 1")
plt.plot(points, f2(points), label="Convex 2")
plt.plot(points, f3(points), label="Convex 3")

plt.plot(points, g1(points), '--', label="Non-Convex 1")
plt.plot(points, g2(points), '--', label="Non-Convex 2")
plt.plot(points, g3(points), '--', label="Non-Convex 3")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Convex vs Non-Convex Functions")
plt.legend()
plt.grid(True)
plt.show()
```
