```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def func_1(x):
    return 3 * x**2 + 12 * x + 7

def func_2(x):
    return 2 * x**4 - 8 * x**2 + 5
```

```python
def check_curvature(f, points, h=1e-4):
    vals = np.array([
        (f(x + h) - 2 * f(x) + f(x - h)) / (h * h)
        for x in points
    ])

    if np.all(vals > 0):
        return "Strictly Convex"
    if np.all(vals < 0):
        return "Strictly Concave"
    if np.any(vals > 0) and np.any(vals < 0):
        return "Non-Convex"
    
    return "Flat"
```

```python
points = np.linspace(-5, 5, 201)
print("Curvature f1:", check_curvature(func_1, points))
print("Curvature f2:", check_curvature(func_2, points))
```

```python
plt.plot(points, func_1(points), label = "Convex")
plt.plot(points, func_2(points), label = "Non-Convex")
plt.legend()
plt.title("Curvature")
plt.show()
```
